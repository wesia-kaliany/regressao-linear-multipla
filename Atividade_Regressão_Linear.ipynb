```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
```

```python
df1 = pd.DataFrame({
    'x1': [68,45,91,48,47,66,50,52,49,38,88,73],
    'x2': [1.7,1.6,1.8,1.6,1.7,1.8,1.7,1.7,1.7,1.6,1.8,1.7],
    'y':  [174,164,244,154,182,208,163,145,145,137,242,191]
})

X = sm.add_constant(df1[['x1', 'x2']])
modelo1 = sm.OLS(df1['y'], X).fit()
print(modelo1.summary())
```

```python
novo = pd.DataFrame({'const': [1], 'x1': [60], 'x2': [1.7]})
print("Vendas estimadas:", modelo1.predict(novo)[0])
```

```python
df2 = pd.DataFrame({
    'TV': [5.0,2.0,4.0,2.5,3.0,3.5,2.5,3.0],
    'J':  [1.5,2.0,1.5,2.5,3.3,2.3,4.2,2.5],
    'Y':  [96,90,95,92,95,94,94,94]
})

X2 = sm.add_constant(df2[['TV', 'J']])
modelo2 = sm.OLS(df2['Y'], X2).fit()
print(modelo2.summary())
```

```python
novo2 = pd.DataFrame({'const': [1], 'TV': [4.0], 'J': [2.0]})
print("Faturamento estimado (US$ mil):", modelo2.predict(novo2)[0])
```
